```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
mail_data = pd.read_csv('/content/mail_data.csv')
mail_data.head()
```

```python
mail_data.shape
```

```python
mail_data.describe()
```

```python
mail_data.isnull().sum()
```

```python
mail_data.replace({'Category':{'ham':0,'spam':1}},inplace=True)
mail_data.head()
```

Split data

```python
X= mail_data.drop(['Category'],axis=1)
Y= mail_data['Category']
```

```python
print(X)
```

```python
print(Y)
```

data text-->int64

```python
#vicotorization of message
vectorizer = TfidfVectorizer(min_df=1,stop_words='english',lowercase=True)
X_values=X['Message'].values
X=vectorizer.fit_transform(X_values)
print(X.shape)
```

```python
print(X)
```

```python
#split data into test and train
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=3)
print(X.shape,X_train.shape,X_test.shape)
```

Train data --->logistic regression

```python
model = LogisticRegression()
model.fit(X_train,Y_train)
```

Evaluation -->accuracy score

```python
#train data accuracy score
train_pred = model.predict(X_train)
train_acc = accuracy_score(Y_train,train_pred)
print(train_acc)
```

```python
#train test accuracy score
test_pred = model.predict(X_test)
test_acc = accuracy_score(Y_test,test_pred)
print(test_acc)
```

Build a predictive system

```python
email_text="you win a reward with us of value $2000..call on 1234567890 to get it"
email_numerical=vectorizer.transform([email_text])
pred=model.predict(email_numerical)
if pred[0]==0:
  print("ham mail")
else:
  print("spam mail")
```

```python
X_test_exemple = X_test[1114]
pred = model.predict(X_test_exemple)
if pred[0]==0:
  print("ham mail")
else:
  print("spam mail")
```
